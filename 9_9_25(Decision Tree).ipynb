```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 2. Importing dataset
data = pd.read_csv('./Dataset/Social_Network_Ads.csv')
dataset = data
dataset
```

```python
X = dataset.iloc[:,2:-1].values
Y = dataset.iloc[:,-1].values
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X = sc.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.25,random_state=0)
```

```python
from sklearn.tree import DecisionTreeClassifier
classifire = DecisionTreeClassifier(criterion='entropy',random_state=0)
classifire.fit(X_train,Y_train)
```

```python
y_predict = classifire.predict(X_test)
y_predict
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(Y_test,y_predict)
cm
```

```python
from sklearn.metrics import classification_report
print(classification_report(Y_test,y_predict))
```

```python
from sklearn import tree
tree.plot_tree(classifire)
```

```python
accuracy = (cm[0,0]+cm[1,1])/(cm[0,0]+cm[0,1]+cm[1,0]+cm[1,1])
print('accuracy of the model ',accuracy)

precition = (cm[0,0])/(cm[0,0]+cm[1,0])
print('precition of the model ',precition)

recall = (cm[0,0])/(cm[0,0]+cm[0,1])
print('recall of the model ',recall)

f1_score = 2/((1/precition)+(1/recall))
print('F1 score of the model ',f1_score)

specifi = cm[1,1]/(cm[1,1]+cm[0,1])
print('specifiti of the model ',specifi)
```
